## Objetivo general

Esta notebook funciona como unos apuntes para introducir el uso de Linux en bioinformática. Su objetivo principal es aprender a trabajar en el terminal, moverse por el sistema de archivos, utilizar los comandos básicos y ejecutar programas desde la línea de comandos.

## Por qué Linux

Linux es el sistema operativo dominante en servidores, clusters, estaciones de trabajo científicas y muchos entornos de computación en la nube. En bioinformática se usa de forma constante porque:

- facilita automatizar tareas repetitivas
- permite procesar grandes volúmenes de ficheros
- es el entorno natural de muchas herramientas bioinformáticas
- favorece la reproducibilidad mediante scripts y registros de comandos
- mejora la accesibilidad porque muchas herramientas son libres y estan disponibles sin costes de licencia

## Reproducibilidad y accesibilidad

Una de las grandes ventajas de Linux y de la terminal es que casi todo lo que haces puede escribirse como una secuencia de comandos. Eso permite volver a ejecutar un análisis, compartirlo con otra persona, revisarlo y adaptarlo a otros datos. Cualquier secuencia de acciones que se realiza manualmente a través de una interfaz de usuario es menos reproducible y automatizable.

## 2. Estructura del sistema de archivos

En Linux, los ficheros y directorios se organizan en forma de arbol. Todo cuelga de una raiz `/`. Para moverte y entender tu ubicacion actual es necesario dominar unos pocos comandos básicos.

### Comandos principales

- `pwd`: muestra el directorio actual.
- `ls`: lista el contenido de un directorio.
- `cd`: cambia de directorio.
- `mkdir`: crea directorios.
- `rmdir`: elimina directorios vacios.
- `tree`: muestra la estructura en forma de arbol.

### Argumentos y rutas importantes

- `.`: directorio actual.
- `..`: directorio padre.
- `~`: directorio personal del usuario.
- `ls -l`: listado largo.
- `ls -a`: incluye elementos ocultos.
- `ls -lh`: tamaños legibles.
- `mkdir -p`: crea directorios intermedios si no existen.

## 3. Gestion de ficheros

En bioinformática se trabaja continuamente con FASTA, FASTQ, GFF, CSV y ficheros comprimidos. Por eso es esencial manejar bien los comandos de copia, movimiento, visualizacion y conteo.

- `mv`: mueve o renombra ficheros o directorios.
- `rm`: elimina ficheros o directorios.
- `touch`: crea un fichero vacio o actualiza su fecha.
- `cat`: muestra el contenido completo de un fichero.

## 4. Permisos, ayuda y documentación

En Linux, cada fichero tiene permisos asociados. Los permisos mas importantes son:

- `r`: lectura
- `w`: escritura
- `x`: ejecucion

Estos permisos se asignan a tres grupos:

- usuario propietario
- grupo
- otros usuarios

Cuando ves algo como `-rw-r--r--`, significa:

- `rw-`: el propietario puede leer y escribir
- `r--`: el grupo solo puede leer
- `r--`: el resto solo puede leer

### Códigos numericos de permisos

- `4`: lectura
- `2`: escritura
- `1`: ejecucion

Por ejemplo:

- `644`: propietario `rw`, grupo `r`, otros `r`
- `755`: propietario `rwx`, grupo `rx`, otros `rx`
- `700`: propietario `rwx`, nadie mas

### Comandos principales

- `chmod`: cambia permisos.
- `man`: abre el manual de un comando.
- `comando --help`: muestra ayuda rapida.
- `history`: muestra el historial de comandos ejecutados.

```python
%%bash
cd ../../
# Ver permisos y cambiarlos en un fichero de prueba
touch data/processed/linux_practica/script_prueba.sh
ls -l data/processed/linux_practica/script_prueba.sh
chmod 755 data/processed/linux_practica/script_prueba.sh
ls -l data/processed/linux_practica/script_prueba.sh
```

```python
%%bash

# Ayuda rapida
ls --help | head -n 10
wc --help | head -n 10
history | tail -n 10
```

`man` abre una pagina interactiva de manual, muy uútil en terminal real. En notebook puede no ser cómodo, pero sigue siendo una de las herramientas más importantes para aprender Linux.

## 5. Redirecciones y pipes

Las redirecciones y pipes son uno de los rasgos más potentes de la shell.

### Operadores principales

- `>`: redirige la salida a un fichero, sobrescribiéndolo.
- `>>`: redirige la salida añadiéndola al final del fichero.
- `<`: usa un fichero como entrada.
- `|`: envía la salida de un comando como entrada del siguiente.

Con esto puedes construir flujos de trabajo donde cada herramienta hace una parte del proceso.

```python
%%bash
cd ../../
# Sobrescribir y anadir contenido
echo "Linea 1" > data/processed/linux_practica/redireccion.txt
echo "Linea 2" >> data/processed/linux_practica/redireccion.txt
cat data/processed/linux_practica/redireccion.txt
```

```python
%%bash
cd ../../
# Pipe basico
cat data/raw/taxonomy_GTDB.csv | head -n 5
```

## 6. Herramientas de procesamiento de texto

Estas herramientas son esenciales en bioinformática porque buena parte de los datos se almacenan como texto estructurado.

### Comandos principales

- `grep`: busca líneas que contienen un patrón.
- `sort`: ordena líneas.
- `uniq`: colapsa líneas repetidas consecutivas.
- `cut`: extrae columnas o rangos de caracteres.
- `tr`: transforma caracteres.
- `paste`: une líneas o columnas.
- `sed`: editor de flujo para sustituciones y transformaciones simples.
- `awk`: lenguaje pequeno orientado a columnas y patrones.

### Argumentos y usos frecuentes

- `grep "patron" fichero`: busca un patrón.
- `grep -c`: cuenta coincidencias.
- `grep -v`: excluye coincidencias.
- `sort -u`: ordena y elimina duplicados.
- `cut -f 1 -d ','`: extrae la primera columna usando coma como separador.
- `tr 'a-z' 'A-Z'`: convierte a mayúsculas.
- `sed 's/viejo/nuevo/'`: sustituye la primera ocurrencia del texto 'viejo' por el 'nuevo' en cada línea.
- `sed 's/viejo/nuevo/g'`: sustituye todas las ocurrencias del texto.
- `awk -F ',' '{print $1, $2}'`: usa coma como separador y muestra columnas.

```python
%%bash
cd ../../
# Contar secuencias en un FASTA buscando cabeceras
grep -c '^>' data/raw/proteinas_bacteria.faa
grep -c '^>' data/raw/genes_bacteria.ffn
```

```python
%%bash
cd ../../
# Extraer columnas de un CSV
head -n 5 data/raw/taxonomy_GTDB.csv | cut -d ',' -f 1,3,8
```

```python
%%bash
cd ../../
# Filtrar genes o anotaciones por patron
grep 'DNA primase' data/raw/anotacion.gff3 | head -n 3

# En un FASTA, las secuencias suelen empezar en una línea que empieza por `>`. 
grep '^>' data/raw/proteinas_bacteria.faa | grep 'Tra' | head -n 5
```

```python
%%bash
cd ../../
# Sort y uniq
cut -d ',' -f 3 data/raw/taxonomy_GTDB.csv | sort | uniq | head
```

```python
%%bash

# tr para convertir texto
echo 'atgcgatcga' | tr 'a-z' 'A-Z'
```

```python
%%bash

# paste para unir columnas de dos salidas
head -n 5 data/raw/taxonomy_GTDB.csv | cut -d ',' -f 1 > data/processed/linux_practica/ident.txt
head -n 5 data/raw/taxonomy_GTDB.csv | cut -d ',' -f 8 > data/processed/linux_practica/genus.txt
paste data/processed/linux_practica/ident.txt data/processed/linux_practica/genus.txt
```

```python
%%bash

# sed minimo: sustitucion simple
echo 'geneA geneB geneC' | sed 's/gene/gen/'
```

```python
%%bash

# awk minimo: extraer columnas y contar campos
head -n 5 data/raw/taxonomy_GTDB.csv | awk -F ',' '{print $1, $3, $8}'
```

## 7. Compresión y archivos grandes

En bioinformática es habitual trabajar con archivos enormes, por lo que la compresion es parte del flujo de trabajo diario.

### Comandos principales

- `gzip`: comprime un fichero.
- `gunzip`: descomprime un fichero `.gz`.
- `zcat`: muestra el contenido de un `.gz` sin descomprimirlo.
- `tar`: empaqueta varios ficheros o directorios en un unico archivo.

### Argumentos frecuentes de `tar`

- `tar -cf archivo.tar directorio`: crea un tar.
- `tar -tf archivo.tar`: lista su contenido.
- `tar -xf archivo.tar`: extrae su contenido.

```python
%%bash
cd ../../
# Trabajar con gzip y zcat
cp data/raw/anotacion_resumen.txt data/processed/linux_practica/anotacion_resumen_tmp.txt
gzip -c data/processed/linux_practica/anotacion_resumen_tmp.txt > data/processed/linux_practica/anotacion_resumen_tmp.txt.gz
zcat data/processed/linux_practica/anotacion_resumen_tmp.txt.gz | head
```

```python
%%bash
cd ../../
# Crear un archivo tar con varios ficheros de prueba
tar -cf data/processed/linux_practica/ejemplo.tar data/processed/linux_practica/nota.txt data/processed/linux_practica/redireccion.txt
tar -tf data/processed/linux_practica/ejemplo.tar
```

## 8. Variables y bash basico

Una variable en bash se asigna sin espacios alrededor del signo igual.

- `VAR=valor`
- `echo $VAR`

Los scripts permiten guardar una secuencia de comandos en un fichero ejecutable.

```python
%%bash

# Variables simples
RUTA_FASTQ=data/raw/pacbio_1.fastq
echo $RUTA_FASTQ
file $RUTA_FASTQ
```

```python
%%bash
cd ../../
# Pequeño bucle en bash
for file in data/raw/*.fastq*
do
  echo $file
done
```

```python
%%bash
cd ../../
# Ejemplo de script sencillo
cat > data/processed/linux_practica/listar_fastq.sh <<'EOF'
#!/bin/bash
for file in data/raw/*.fastq*
do
  echo $file
done
EOF

chmod 755 data/processed/linux_practica/listar_fastq.sh
bash data/processed/linux_practica/listar_fastq.sh
```

## 9. Editores de texto en terminal

Saber moverse mínimamente en un editor de terminal es importante para editar scripts, configuraciones o pequeños ficheros en servidores remotos.

### `nano`

`nano` es un editor sencillo y amable para principiantes. Permite editar texto de forma directa y muestra atajos en la parte inferior de la pantalla.

Uso básico:

- `nano archivo.txt`

### `vim`

`vim` es mucho más potente, pero también menos intuitivo al principio. Conviene al menos conocer lo minimo para no quedarse bloqueado.

Comandos minimos de `vim`:

- `i`: entra en modo inserción para escribir
- `ESC`: vuelve al modo normal
- `:wq`: guarda y sale
- `:q!`: sale sin guardar

No es necesario dominar `vim` al completo, pero sí saber entrar, editar un archivo, guardarlo y salir.

```python
%%bash
cd ../../
# Ejemplos de uso, para ejecutar en terminal interactiva real
echo "nano data/processed/linux_practica/nota.txt"
echo "vim data/processed/linux_practica/nota.txt"
```
