## 1. Tipos compuestos: vision general

- `list`: secuencia ordenada y mutable.

## 2. Listas: creacion, acceso y métodos principales

Las listas son secuencias ordenadas y mutables. Admiten elementos repetidos y de tipos distintos, aunque en la práctica suele convenir que sean homogeneas.

```python
# Ejemplo 2: creacion y acceso
genes = ["BRCA1", "TP53", "EGFR", "MYC"]
print(genes)
print(genes[0])
print(genes[-1])
print(len(genes))
```

```python
# Ejemplo 3: modificacion
genes[1] = "PTEN"
print(genes)
```

```python
# Ejemplo 4: append e insert
genes.append("KRAS")
genes.insert(1, "BRAF")
print(genes)
```

```python
# Ejemplo 5: extend
genes.extend(["ALK", "RET"])
print(genes)
```

```python
# Ejemplo 6: remove y pop
genes.remove("ALK")
eliminado = genes.pop()
print(genes)
print(eliminado)
```

```python
# Ejemplo 7: count e index
bases = ["A", "T", "G", "C", "A", "G", "A"]
print(bases.count("A"))
print(bases.index("G"))
```

```python
# Ejemplo 8: sort, sorted y reverse
valores = [8, 3, 10, 1, 7]
print(sorted(valores))
valores.sort()
print(valores)
valores.reverse()
print(valores)
```

```python
# Ejemplo 9: slicing
secuencia = ["A", "T", "G", "C", "G", "A"]
print(secuencia[1:4])
print(secuencia[:3])
print(secuencia[::2])
```

## 4. Sets

Los sets almacenan elementos unicos. Son utiles para eliminar duplicados y para operaciones de conjunto.

```python
# Ejemplo 12: crear sets y eliminar duplicados
muestras = ["M1", "M2", "M2", "M3", "M1"]
muestras_unicas = set(muestras)
print(muestras_unicas)
```

```python
# Ejemplo 13: add, discard y remove
genes = {"BRCA1", "TP53"}
genes.add("EGFR")
genes.discard("MYC") #Si MYC está presente, lo elimina; si no está, no pasa nada
print(genes)
genes.remove("TP53") #Si TP53 está presente, lo elimina; si no está, salta un error
print(genes)
```

```python
# Ejemplo 14: operaciones de conjunto
a = {"BRCA1", "TP53", "EGFR"}
b = {"TP53", "MYC", "EGFR"}
print(a.union(b))
print(a.intersection(b))
print(a.difference(b))
print(a.symmetric_difference(b))
```

```python
# Ejemplo 15: subset y superset
x = {1, 2}
y = {1, 2, 3, 4}
print(x.issubset(y))
print(y.issuperset(x))
```

## 5. Diccionarios: creacion, acceso y métodos principales

Los diccionarios relacionan claves con valores. Son muy utiles para almacenar anotaciones, mediciones o metadatos.

```python
# Ejemplo 16: crear y consultar diccionarios
expresion = {"BRCA1": 12.4, "TP53": 8.2, "EGFR": 15.7}
print(expresion)
print(expresion["BRCA1"])
print(len(expresion))
```

```python
# Ejemplo 17: añadir y modificar valores
expresion["MYC"] = 21.0
expresion["TP53"] = 9.1
print(expresion)
```

```python
# Ejemplo 18: get
print(expresion.get("EGFR"))
print(expresion.get("ALK"))
print(expresion.get("ALK", 0.0))
```

```python
# Ejemplo 19: keys, values, items
print(expresion.keys())
print(expresion.values())
print(expresion.items())
```

```python
# Ejemplo 20: pop, popitem y update
valor = expresion.pop("MYC")
print(valor)
print(expresion)
expresion.update({"KRAS": 7.5, "BRAF": 11.2})
print(expresion)
ultimo = expresion.popitem()
print(ultimo)
print(expresion)
```

```python
# Ejemplo 21: clear y copy
expresion_copia = expresion.copy()
print(expresion_copia)
temporal = {"A": 1, "B": 2}
temporal.clear()
print(temporal)
```

## 7. Referencia, copia, `is` e `is not`

Con tipos compuestos, una asignación no siempre crea una copia nueva. A menudo crea una segunda referencia al mismo objeto.

```python
# Ejemplo 24: referencia compartida
lista1 = ["A", "T", "G"]
lista2 = lista1
lista2.append("C")
print(lista1)
print(lista2)
print(lista1 is lista2)
```

```python
# Ejemplo 25: copia superficial
lista3 = ["A", "T", "G"]
lista4 = lista3.copy()
lista4.append("C")
print(lista3)
print(lista4)
print(lista3 is lista4)
print(lista3 is not lista4)
```

```python
# Ejemplo 26: igualdad frente a identidad
a = [1, 2, 3]
b = [1, 2, 3]
print(a == b)
print(a is b)
```

## 11. List comprehensions y dict comprehensions

```python
# Ejemplo 39: list comprehension basica
cuadrados = [x ** 2 for x in range(6)]
print(cuadrados)
```

```python
# Ejemplo 40: list comprehension con condicion
pares = [x for x in range(10) if x % 2 == 0]
print(pares)
```

## 12. Busqueda, clasificación y ordenacion

```python
# Ejemplo 43: buscar un elemento manualmente
genes = ["BRCA1", "TP53", "EGFR", "MYC"]
objetivo = "EGFR"
encontrado = False

for gen in genes:
    if gen == objetivo:
        encontrado = True
        break

print(encontrado)
```

```python
# Ejemplo 44: clasificar valores
expresiones = [2.1, 8.3, 12.4, 1.0, 15.2]
altas = []
bajas = []

for valor in expresiones:
    if valor >= 10:
        altas.append(valor)
    else:
        bajas.append(valor)

print(altas)
print(bajas)
```

```python
# Ejemplo 45: ordenar una lista de tuplas
datos = [("BRCA1", 12.4), ("TP53", 8.2), ("EGFR", 15.7)]
datos.sort(key=lambda x: x[1])
print(datos)
```

## 14. Problemas integrados con listas, diccionarios, sets y tuplas

```python
# Ejemplo 48: construir un diccionario a partir de dos listas
genes = ["BRCA1", "TP53", "EGFR"]
valores = [12.4, 8.2, 15.7]
resultado = {}

for i in range(len(genes)):
    resultado[genes[i]] = valores[i]

print(resultado)
```

```python
# Ejemplo 49: eliminar duplicados manteniendo la idea general
genes = ["BRCA1", "TP53", "BRCA1", "EGFR", "TP53"]
genes_unicos = list(set(genes))
print(genes_unicos)
```

```python
# Ejemplo 50: Fibonacci como lista
n = 8
fibonacci = [0, 1]

while len(fibonacci) < n:
    fibonacci.append(fibonacci[-1] + fibonacci[-2])

print(fibonacci)
```

```python
# Ejemplo 51: clasificar genes por expresion
expresion = {"BRCA1": 12.4, "TP53": 8.2, "EGFR": 15.7, "MYC": 4.1}
altos = []
bajos = []

for gen, valor in expresion.items():
    if valor >= 10:
        altos.append(gen)
    else:
        bajos.append(gen)

print(altos)
print(bajos)
```

## 15. `zip()` y funciones afines

La función `zip()` permite recorrer varias secuencias a la vez y agrupar sus elementos por posicion. Es muy útil cuando tenemos listas paralelas, por ejemplo genes y valores de expresión.

```python
# Ejemplo 52: zip basico
genes = ["BRCA1", "TP53", "EGFR"]
valores = [12.4, 8.2, 15.7]

for gen, valor in zip(genes, valores):
    print(gen, valor)
```

```python
# Ejemplo 53: convertir zip en lista de tuplas
pares = list(zip(genes, valores))
print(pares)
```

```python
# Ejemplo 54: construir un diccionario con zip
expresion = dict(zip(genes, valores))
print(expresion)
```

```python
# Ejemplo 55: zip con enumerate
for i, (gen, valor) in enumerate(zip(genes, valores), start=1):
    print(i, gen, valor)
```

## 17. Cadenas de texto, patrones e iteradores

En bioinformatica y análisis de datos es muy habitual recorrer cadenas caracter a caracter, buscar motivos o patrones y usar expresiones regulares para reconocer formatos o secuencias concretas.

```python
# Ejemplo 56: iterar sobre una cadena
secuencia = "ATGCGTAA"
for base in secuencia:
    print(base)
```

```python
# Ejemplo 57: buscar patrones sencillos
secuencia = "ATGCGTAA"
print("ATG" in secuencia)
print(secuencia.find("CGT"))
print(secuencia.count("A"))
print(secuencia.startswith("ATG"))
print(secuencia.endswith("TAA"))
```

```python
# Ejemplo 58: expresiones regulares
import re

texto = "Muestras: M1, M2, M15, ControlA"
coincidencias = re.findall(r"M\d+", texto)
print(coincidencias)
```
