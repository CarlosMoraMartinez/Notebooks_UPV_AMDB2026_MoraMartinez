## 1. Condicionales `if`, `elif`, `else`

Las estructuras condicionales permiten ejecutar codigo distinto segun se cumpla o no una condicion booleana.

```python
# Ejemplo 1: if simple
longitud = 1500

if longitud > 1000:
    print("La secuencia es larga")
```

```python
# Ejemplo 2: if-else
temperatura = 37

if temperatura == 37:
    print("Temperatura fisiologica")
else:
    print("Temperatura no fisiologica")
```

```python
# Ejemplo 3: if-elif-else
ph = 7.4

if ph < 7:
    print("Medio acido")
elif ph == 7:
    print("Medio neutro")
else:
    print("Medio basico")
```

## 2. Operadores logicos: `and`, `or`, `not`

Los operadores logicos permiten combinar o invertir condiciones:

- `and`: ambas condiciones deben ser verdaderas.
- `or`: basta con que una condicion sea verdadera.
- `not`: invierte el valor booleano.

```python
# Ejemplo 4: operador and
temperatura = 37
ph = 7.2

if temperatura == 37 and ph > 7:
    print("Condiciones favorables")
```

```python
# Ejemplo 5: operador or
gen = "TP53"

if gen == "TP53" or gen == "BRCA1":
    print("Gen de interes prioritario")
```

```python
# Ejemplo 6: operador not
muestra_contaminada = False

if not muestra_contaminada:
    print("La muestra puede procesarse")
```

## 3. Bucles `for` y `range()`

Un bucle `for` repite un bloque de codigo un numero determinado de veces o recorriendo elementos.

`range()` puede usarse como:

- `range(fin)`
- `range(inicio, fin)`
- `range(inicio, fin, paso)`

```python
# Ejemplo 7: contar con for
for i in range(5):
    print(i)
```

```python
# Ejemplo 8: range con inicio y fin
for ciclo in range(1, 6):
    print("Ciclo", ciclo)
```

```python
# Ejemplo 9: range con paso
for temperatura in range(20, 41, 5):
    print(temperatura)
```

```python
# Ejemplo 10: sumatorio
total = 0
for numero in range(1, 6):
    total += numero
print(total)
```

## 4. Bucles `while`

Un bucle `while` es apropiado cuando no sabemos exactamente cuantas iteraciones necesitaremos, sino que queremos repetir mientras se cumpla una condicion.

```python
# Ejemplo 11: while sencillo
contador = 0

while contador < 5:
    print(contador)
    contador += 1
```

```python
# Ejemplo 12: sumatorio con while
n = 1
total = 0

while n <= 5:
    total += n
    n += 1

print(total)
```

```python
# Ejemplo 13: caso apropiado para while
concentracion = 100

while concentracion > 10:
    print(concentracion)
    concentracion -= 15
```

## 5. `break`, `continue` y `pass`

- `break` termina el bucle inmediatamente.
- `continue` salta a la siguiente iteracion.
- `pass` no hace nada y se usa como marcador temporal.

```python
# Ejemplo 14: break
for numero in range(10):
    if numero == 4:
        break
    print(numero)
```

```python
# Ejemplo 15: continue
for numero in range(6):
    if numero == 3:
        continue
    print(numero)
```

```python
# Ejemplo 16: pass
for numero in range(3):
    if numero == 1:
        pass
    print(numero)
```

## 6. Estructuras anidadas

Podemos anidar condicionales dentro de bucles, bucles dentro de condicionales y bucles dentro de otros bucles.

```python
# Ejemplo 17: if dentro de for
for numero in range(1, 6):
    if numero % 2 == 0:
        print(numero, "es par")
    else:
        print(numero, "es impar")
```

```python
# Ejemplo 18: for dentro de for
for fila in range(1, 4):
    for columna in range(1, 3):
        print("Fila", fila, "Columna", columna)
```

```python
# Ejemplo 19: while con if
contador = 1

while contador <= 5:
    if contador == 3:
        print("Punto medio")
    else:
        print(contador)
    contador += 1
```

## 7. Problemas integrados clasicos

En este bloque se combinan varias herramientas de control de flujo.

```python
# Ejemplo 20: factorial con for
n = 5
factorial = 1

for i in range(1, n + 1):
    factorial *= i

print(factorial)
```

```python
# Ejemplo 21: mostrar n numeros
n = 7

for i in range(1, n + 1):
    print(i)
```

```python
# Ejemplo 22: suma de pares hasta n
n = 10
total = 0

for i in range(1, n + 1):
    if i % 2 == 0:
        total += i

print(total)
```
