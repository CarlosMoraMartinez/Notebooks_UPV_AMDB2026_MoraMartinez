Una variable es un nombre que referencia un valor. En Python no se puede usar una variable antes de asignarle un valor.

Python no tiene constantes estrictas, pero por convenio se escriben en mayusculas.

```python
# Ejemplo 1: asignacion sencilla
gen = "TP53"
print(gen)
```

```python
# Ejemplo 2: reasignacion
lecturas = 120
print(lecturas)
lecturas = 145
print(lecturas)
```

```python
# Ejemplo 3: constante por convencion
PI = 3.14159
MAX_MUESTRAS = 96
print(PI)
print(MAX_MUESTRAS)
```

## 6. Cadenas de texto: declaracion, ASCII, Unicode y escapes

En Python las cadenas pueden declararse con comillas simples, comillas dobles o triples comillas.

ASCII representa un conjunto basico de caracteres. Unicode extiende ese conjunto para incluir muchisimos alfabetos, simbolos y caracteres especiales. En Python 3 las cadenas son Unicode, lo que facilita trabajar con nombres cientificos, acentos o simbolos.

Escapes principales:

- `\n`: salto de linea
- `\t`: tabulador
- `\r`: retorno de carro

```python
# Ejemplo 13: distintas formas de declarar cadenas
cadena1 = 'ADN'
cadena2 = "ARN"
cadena3 = '''Proteina\nmultilinea'''
print(cadena1)
print(cadena2)
print(cadena3)
```

```python
# Ejemplo 14: caracteres Unicode
nombre = "José"
especie = "Caenorhabditis elegans"
simbolo = "β-globina"
print(nombre)
print(especie)
print(simbolo)
```

```python
# Ejemplo 15: escapes
print("Gen\tExpresion")
print("BRCA1\t12.4")
print("Linea 1\nLinea 2")
print("ABC\rXYZ")
```

## 7. Interpolacion y composicion de cadenas

Formas frecuentes de interpolar valores en cadenas:

- concatenacion con `+`
- separando argumentos en `print()`
- estilo antiguo con `%`
- metodo `.format()`
- f-strings

```python
# Ejemplo 16: concatenacion
gen = "MYC"
mensaje = "El gen analizado es " + gen
print(mensaje)
```

```python
# Ejemplo 17: print con varios argumentos
gen = "EGFR"
expresion = 8.45
print("Gen:", gen, "Expresion:", expresion)
```

```python
# Ejemplo 18: interpolacion con %
gen = "PTEN"
valor = 5.2
print("El gen %s tiene expresion %.1f" % (gen, valor))
```

```python
# Ejemplo 19: metodo format
gen = "KRAS"
mutacion = "G12D"
print("El gen {} presenta la mutacion {}".format(gen, mutacion))
print("La mutacion {m} pertenece al gen {g}".format(g=gen, m=mutacion))
```

```python
# Ejemplo 20: f-strings
gen = "BRAF"
frecuencia = 0.2378
print(f"El gen {gen} presenta una frecuencia de {frecuencia:.2%}")
```
