## 3. Constructor `__init__`

El constructor `__init__` se ejecuta automáticamente al crear un objeto. Suele usarse para inicializar atributos, es decir, variables asociadas al objeto.

```python
# Ejemplo 4: clase con constructor
class Gen:
    def __init__(self, nombre, longitud):
        self.nombre = nombre
        self.longitud = longitud

gen1 = Gen("BRCA1", 1863)
print(gen1.nombre)
print(gen1.longitud)
```

## 4. Metodos

Un metodo es una funcion definida dentro de una clase. Suele trabajar con los atributos del objeto y por eso recibe `self`, que representa al propio objeto.

```python
# Ejemplo 5: clase con metodo
class Muestra:
    def __init__(self, identificador, tejido):
        self.identificador = identificador
        self.tejido = tejido

    def describir(self):
        print(f"Muestra {self.identificador} del tejido {self.tejido}")

m = Muestra("M1", "higado")
m.describir()
```

```python
# Ejemplo 6: metodo que devuelve un valor
class Secuencia:
    def __init__(self, texto):
        self.texto = texto

    def longitud(self):
        return len(self.texto)

seq = Secuencia("ATGCGT")
print(seq.longitud())
```

## 5. Clases sencillas con varios metodos

```python
# Ejemplo 7: varios metodos sencillos
class Gen:
    def __init__(self, nombre, secuencia):
        self.nombre = nombre
        self.secuencia = secuencia

    def longitud(self):
        return len(self.secuencia)

    def tiene_inicio_atg(self):
        return self.secuencia.startswith("ATG")

gen = Gen("genA", "ATGCGTAA")
print(gen.longitud())
print(gen.tiene_inicio_atg())
```

## 6. Metodos que modifican atributos

Los metodos no solo pueden consultar datos. Tambien pueden modificar los atributos del objeto.

```python
# Ejemplo 8: modificar un atributo
class Contador:
    def __init__(self, valor):
        self.valor = valor

    def incrementar(self):
        self.valor += 1

c = Contador(0)
c.incrementar()
c.incrementar()
print(c.valor)
```

```python
# Ejemplo 9: cambiar una secuencia
class SecuenciaSimple:
    def __init__(self, secuencia):
        self.secuencia = secuencia

    def convertir_a_mayusculas(self):
        self.secuencia = self.secuencia.upper()

s = SecuenciaSimple("atgc")
s.convertir_a_mayusculas()
print(s.secuencia)
```

## 7. Relacion con listas y otros objetos conocidos

Cuando llamas a `append()` sobre una lista o a `upper()` sobre una cadena, ya estas usando objetos y metodos. Al definir tus propias clases estas creando tus propios tipos de objeto.

```python
# Ejemplo 10: lista de objetos
class Muestra:
    def __init__(self, identificador):
        self.identificador = identificador

muestras = [Muestra("M1"), Muestra("M2"), Muestra("M3")]
for muestra in muestras:
    print(muestra.identificador)
```

## 8. Introduccion teorica a la herencia

La herencia permite definir una clase hija a partir de una clase padre. La clase hija hereda atributos y metodos y puede añadir otros nuevos o modificar algunos. En esta practica solo veremos la idea general, sin profundizar demasiado.

```python
# Ejemplo 11: ejemplo muy basico de herencia
class Organismo:
    def __init__(self, nombre):
        self.nombre = nombre

    def describir(self):
        print(f"Organismo: {self.nombre}")

class Bacteria(Organismo):
    def decir_tipo(self):
        print("Soy una bacteria")

b = Bacteria("E. coli")
b.describir()
b.decir_tipo()
```
