## 5. Funciones numericas e integracion con listas

En este bloque las funciones se usan para resolver calculos y operaciones sobre colecciones de datos. La idea es pasar de ejemplos muy simples a pequeñas herramientas reutilizables.

```python
# Ejemplo 9: media de una lista
def media_lista(valores):
    return sum(valores) / len(valores)

print(media_lista([2, 4, 6, 8]))
```

```python
# Ejemplo 10: filtrar valores por umbral
def filtrar_mayores(valores, umbral):
    salida = []
    for valor in valores:
        if valor > umbral:
            salida.append(valor)
    return salida

print(filtrar_mayores([1, 5, 12, 7, 18], 8))
```

```python
# Ejemplo 11: factorial iterativo
def factorial_iterativo(n):
    resultado = 1
    for i in range(1, n + 1):
        resultado *= i
    return resultado

print(factorial_iterativo(5))
```

```python
# Ejemplo 12: distancia euclidea entre dos vectores
def distancia_euclidea(v1, v2):
    suma = 0
    for a, b in zip(v1, v2):
        suma += (a - b) ** 2
    return suma ** 0.5

print(distancia_euclidea([1, 2], [4, 6]))
```

```python
# Ejemplo 13: area de un poligono regular muy simple
def area_poligono_regular(perimetro, apotema):
    return perimetro * apotema / 2

print(area_poligono_regular(20, 3))
```

## 6. Reutilizacion y combinacion de funciones

Una gran ventaja de las funciones es que pueden llamarse entre si. Esto permite construir tareas complejas a partir de piezas pequeñas y claras.

```python
# Ejemplo 14: una funcion llama a otra
def suma_lista(valores):
    return sum(valores)

def media_desde_suma(valores):
    return suma_lista(valores) / len(valores)

print(media_desde_suma([2, 4, 6]))
```

```python
# Ejemplo 15: usar funciones dentro de un flujo mayor
def es_alta_expresion(valor, umbral=10):
    return valor >= umbral

def clasificar_expresiones(valores, umbral=10):
    altas = []
    bajas = []
    for valor in valores:
        if es_alta_expresion(valor, umbral):
            altas.append(valor)
        else:
            bajas.append(valor)
    return altas, bajas

print(clasificar_expresiones([4, 11, 7, 15, 20]))
```

## 7. Funciones con listas y diccionarios

```python
# Ejemplo 16: contar elementos de una lista en un diccionario
def contar_elementos(lista):
    conteo = {}
    for elemento in lista:
        conteo[elemento] = conteo.get(elemento, 0) + 1
    return conteo

print(contar_elementos(["A", "T", "A", "G", "A", "C"]))
```

```python
# Ejemplo 17: separar pares e impares
def separar_pares_impares(valores):
    pares = []
    impares = []
    for valor in valores:
        if valor % 2 == 0:
            pares.append(valor)
        else:
            impares.append(valor)
    return pares, impares

print(separar_pares_impares([1, 2, 3, 4, 5, 6]))
```

## 8. Funciones para leer ficheros

Las funciones son especialmente utiles para lectura de ficheros, porque el parseo de cada formato puede encapsularse en una herramienta reutilizable.

```python
# Ejemplo 18: leer un fichero de texto y devolver sus lineas limpias
def leer_lineas_limpias(ruta):
    lineas = []
    with open(ruta, "r") as f:
        for linea in f:
            linea = linea.strip()
            if linea:
                lineas.append(linea)
    return lineas

print(leer_lineas_limpias("data/raw/anotacion_resumen.txt")[:4])
```

```python
# Ejemplo 19: leer un resumen clave-valor
def leer_resumen_como_diccionario(ruta):
    resumen = {}
    with open(ruta, "r") as f:
        for linea in f:
            linea = linea.strip()
            if not linea or linea.endswith(":"):
                continue
            clave, valor = linea.split(":", maxsplit=1)
            resumen[clave.strip()] = valor.strip()
    return resumen

print(leer_resumen_como_diccionario("data/raw/anotacion_resumen.txt"))
```

```python
# Ejemplo 20: leer cabeceras de un FASTA
def leer_cabeceras_fasta(ruta):
    cabeceras = []
    with open(ruta, "r") as f:
        for linea in f:
            if linea.startswith(">"):
                cabeceras.append(linea.strip())
    return cabeceras

print(leer_cabeceras_fasta("data/raw/proteinas_bacteria.faa")[:2])
```

## 9. Funciones para escribir ficheros

```python
# Ejemplo 21: escribir una lista en un fichero
def escribir_lista_en_fichero(lista, ruta):
    with open(ruta, "w") as f:
        for elemento in lista:
            f.write(str(elemento) + "\n")

escribir_lista_en_fichero(["A", "B", "C"], "data/processed/ejemplo_lista.txt")
```

```python
# Ejemplo 22: escribir un diccionario simple
def escribir_diccionario(dic, ruta):
    with open(ruta, "w") as f:
        for clave, valor in dic.items():
            f.write(f"{clave}\t{valor}\n")

escribir_diccionario({"A": 1, "B": 2}, "data/processed/ejemplo_diccionario.txt")
```

## 11. `yield` y generadores

Una función con `yield` no devuelve todo de golpe. Va produciendo valores uno a uno. Eso es útil cuando no queremos cargar todos los datos en memoria a la vez o cuando queremos generar secuencias progresivamente.

```python
# Ejemplo 25: generador sencillo
def contar_hasta(n):
    for i in range(1, n + 1):
        yield i

for valor in contar_hasta(5):
    print(valor)
```

```python
# Ejemplo 26: leer secuencias FASTA con yield
def iterar_fasta(ruta):
    ident = None
    secuencia = []
    with open(ruta, "r") as f:
        for linea in f:
            linea = linea.strip()
            if linea.startswith(">"):
                if ident is not None:
                    yield ident, "".join(secuencia)
                ident = linea[1:].split()[0]
                secuencia = []
            else:
                secuencia.append(linea)
        if ident is not None:
            yield ident, "".join(secuencia)

for ident, seq in iterar_fasta("data/raw/proteinas_bacteria.faa"):
    print(ident, len(seq))
    break
```

## 12. Introducción a la recursión

Una función recursiva se llama a si misma. Para que funcione correctamente necesita al menos un caso base, que detenga las llamadas, y un caso recursivo, que reduzca el problema.

```python
# Ejemplo 27: factorial recursivo
def factorial_recursivo(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial_recursivo(n - 1)

print(factorial_recursivo(5))
```

```python
# Ejemplo 28: Fibonacci recursivo
def fibonacci_recursivo(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fibonacci_recursivo(n - 1) + fibonacci_recursivo(n - 2)

print(fibonacci_recursivo(7))
```

```python
# Ejemplo 29: funcion de Ackermann
def ackermann(m, n):
    if m == 0:
        return n + 1
    if m > 0 and n == 0:
        return ackermann(m - 1, 1)
    return ackermann(m - 1, ackermann(m, n - 1))

print(ackermann(2, 2))
```
