## 3. Lectura de un fichero resumen y conversion a diccionario

```python
# Ejemplo 5: convertir un resumen simple en diccionario
resumen = {}

with open("data/raw/anotacion_resumen.txt", "r") as f:
    for linea in f:
        linea = linea.strip()
        if not linea or linea.endswith(":"):
            continue
        clave, valor = linea.split(":", maxsplit=1)
        resumen[clave.strip()] = valor.strip()

print(resumen)
```

## 5. Lectura de FASTA de genes y proteinas

```python
# Ejemplo 8: leer cabeceras de un FASTA
cabeceras = []

with open("data/raw/proteinas_bacteria.faa", "r") as f:
    for linea in f:
        if linea.startswith(">"):
            cabeceras.append(linea.strip())

print(cabeceras[:3])
```

```python
# Ejemplo 9: contar secuencias en un FASTA
contador = 0

with open("data/raw/genes_bacteria.ffn", "r") as f:
    for linea in f:
        if linea.startswith(">"):
            contador += 1

print(contador)
```

```python
# Ejemplo 10: leer un FASTA como diccionario simple id -> secuencia
secuencias = {}
ident = None

with open("data/raw/proteinas_bacteria.faa", "r") as f:
    for linea in f:
        linea = linea.strip()
        if linea.startswith(">"):
            ident = linea[1:].split()[0]
            secuencias[ident] = ""
        else:
            secuencias[ident] += linea

primer_id = list(secuencias.keys())[0]
print(primer_id)
print(len(secuencias[primer_id]))
```

## 6. Lectura de GFF3 y parseo de anotaciones

```python
# Ejemplo 11: ignorar comentarios y separar columnas
with open("data/raw/anotacion.gff3", "r") as f:
    for linea in f:
        if linea.startswith("#"):
            continue
        campos = linea.strip().split("\t")
        print(campos[:5])
        break
```

```python
# Ejemplo 12: contar tipos de feature
tipos = {}

with open("data/raw/anotacion.gff3", "r") as f:
    for linea in f:
        if linea.startswith("#"):
            continue
        campos = linea.strip().split("\t")
        tipo = campos[2]
        tipos[tipo] = tipos.get(tipo, 0) + 1

print(tipos)
```

```python
# Ejemplo 13: extraer atributos de la novena columna
with open("data/raw/anotacion.gff3", "r") as f:
    for linea in f:
        if linea.startswith("#"):
            continue
        campos = linea.strip().split("\t")
        atributos = campos[8].split(";")
        dic_atributos = {}
        for atributo in atributos:
            if "=" in atributo:
                clave, valor = atributo.split("=", maxsplit=1)
                dic_atributos[clave] = valor
        print(dic_atributos)
        break
```

## 7. Lectura de FASTQ y analisis de calidad

En un FASTQ, cada lectura ocupa 4 lineas: cabecera, secuencia, separador `+` y calidad.

```python
# Ejemplo 14: leer una lectura FASTQ completa
with open("data/raw/pacbio_1.fastq", "r") as f:
    header = f.readline().strip()
    seq = f.readline().strip()
    plus = f.readline().strip()
    qual = f.readline().strip()

print(header)
print(len(seq), len(qual))
print(plus)
```

```python
# Ejemplo 15: contar lecturas y longitudes
num_lecturas = 0
longitudes = []

with open("data/raw/pacbio_1.fastq", "r") as f:
    while True:
        header = f.readline().strip()
        if not header:
            break
        seq = f.readline().strip()
        f.readline()
        f.readline()
        num_lecturas += 1
        longitudes.append(len(seq))

print(num_lecturas)
print(longitudes[:5])
```

```python
# Ejemplo 16: filtrar lecturas por longitud minima
ids_largas = []

with open("data/raw/pacbio_1.fastq", "r") as f:
    while True:
        header = f.readline().strip()
        if not header:
            break
        seq = f.readline().strip()
        f.readline()
        f.readline()
        if len(seq) >= 1499:
            ids_largas.append(header)

print(len(ids_largas))
```

## 8. Lectura de FASTQ comprimido con `gzip`

```python
# Ejemplo 17: leer un FASTQ comprimido
import gzip

with gzip.open("data/raw/sample2_R1.fastq.gz", "rt") as f:
    for _ in range(4):
        print(f.readline().strip())
```

```python
# Ejemplo 18: contar lecturas en un FASTQ gz
contador = 0

with gzip.open("data/raw/sample2_R1.fastq.gz", "rt") as f:
    while True:
        header = f.readline().strip()
        if not header:
            break
        f.readline()
        f.readline()
        f.readline()
        contador += 1

print(contador)
```

```python
# Ejemplo 19: filtrar por calidad muy simple usando caracteres
buenas = 0
malas = 0

with gzip.open("data/raw/sample2_R1.fastq.gz", "rt") as f:
    while True:
        header = f.readline().strip()
        if not header:
            break
        seq = f.readline().strip()
        f.readline()
        qual = f.readline().strip()
        if "!" in qual or '"' in qual:
            malas += 1
        else:
            buenas += 1

print(buenas, malas)
```
